# (Filtering)

* Finds orders where Sales are greater than 1000.

```python
df[df["Sales"]>1000]
```

* Find orders sales less than 100 


```python


df[df["Sales"]<1000]
```

* Find all records belonging to technology

```python
df[df["Category"]== "Technology"]
```

* Find records excluding Technology category

```python
df[df["Category"] != "Technology"]
```

* AND Operator -> For Filtering records where both conditions are true
* Find records where category is technology and sales > 1000

```python
df[
    (df["Category"]=="Technology") & (df["Sales"] > 1000)
]
```

* OR Operator
* Find records from technology or Furniture

```python
df[
    (df["Category"]=="Technology") | (df["Category"]=="Furniture")
]
```

* NOT Operator
* Show all orders except those belonging to Furniture.

```python
df[~(df["Category"]=="Furniture")]
```

* Find all orders where Category is either Technology or Furniture. 

```python
df[df["Category"].isin(["Technology", "Furniture"])]  # Works same as OR Operator
```

* Which orders have Sales between 500 and 1,000?

```python
df[df["Sales"].between(500, 1000)]
```

* Find Technology orders where Sales are greater than 1,000. 

```python
df.query("Category == 'Technology' and Sales > 1000")
```

# (Modifying Data)

* How can we rename the Sales column to Total Sales?

```python
df.rename(columns={"Sales":"Total Sales"})
```

* How can we remove the Postal Code column if it isn't required?

```python
df.drop(columns=["Postal Code"])
```

* How can we convert Postal Code to an integer-type column?

```python
df["Postal Code"] = df["Postal Code"].astype("int64")
```

* How can we replace "Consumer" with "Individual" in Segment? 

```python
df["Segment"].replace({
    "Consumer" : "Individual"
})
```

* How can we create shorter labels for the three categories?

```python
df["Category"].map({
    "Technology":"Tech",
    "Furniture":"Home",
    "Office Supplies":"Office"
})
```

* How can we round every Sales value to two decimal places?

```python
df["Sales"].apply(lambda x: round(x,2))
```

* How can we convert all customer names to lowercase? 

```python
df["Customer Name"].str.lower()
```

* How can we convert all country names to uppercase? 

```python
df["Country"].str.upper()
```

* How can we standardize city names using title case? 

```python
df["City"].str.title()  # Use: Converts text to title case.
```

* How can we remove unnecessary spaces from Customer Name?

```python
df["Customer Name"].str.strip()     # Use: Removes leading/trailing whitespace.
```

* Which products contain the word "Chair"?

```python
df[df["Product Name"].str.contains(
    "Chair",
    case = False,
    na = False
    )
]
```
