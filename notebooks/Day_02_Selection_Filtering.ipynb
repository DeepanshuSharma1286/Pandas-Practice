# (GroupBy & Aggregation)

* What is the total Sales generated by each Category?

```python
df.groupby("Category")["Sales"].sum()     # Use: Groups records and performs calculations for each group.
```

* What is the total Sales generated by each Region?

```python
df.groupby("Region")["Sales"].sum()     # Use: Calculates Sales totals for each Region.
```

* What is the average Sales value for each customer Segment?

```python
df.groupby("Segment")["Sales"].mean()       # Use: Calculates average Sales for each Segment.
```

### Multiple-column groupby()
* What is the total Sales for each combination of Category and Sub-Category?

```python
df.groupby(
    ["Category","Sub-Category"]        # Use: Performs grouped analysis at multiple levels.
)["Sales"].sum()
```

* For each Category, what are the total, average, minimum and maximum Sales?

```python
df.groupby("Category")["Sales"].agg(
    ["sum","mean","min","max"]
)
```

### Named aggregation
* Create a Category-level report containing total, average and maximum Sales.

```python
df.groupby("Category").agg(         # Use: Creates a clean aggregated report with meaningful column names.
    Total_Sales = ("Sales", "sum"),
    Average_Sales = ("Sales", "mean"),
    Maximum_Sales = ("Sales","max") 
)
```
